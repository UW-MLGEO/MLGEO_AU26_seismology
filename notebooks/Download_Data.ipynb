```python

import os
import pandas as pd
import h5py
import gdown

# Force path resolution relative to repository root
repo_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
dataset_dir = os.path.join(repo_root, "data", "raw")
os.makedirs(dataset_dir, exist_ok=True)

waveform_file_id = "1SrbiQpBpU6mPq5Un_lJpPcfBekwczLzp"
metadata_file_id = "1HK2AuWPQj3dCdKShYcJ7a5E577XASrab"

waveform_url = f"https://drive.google.com/uc?id={waveform_file_id}"
metadata_url = f"https://drive.google.com/uc?id={metadata_file_id}"

waveform_path = os.path.join(dataset_dir, "waveforms.hdf5")
metadata_path = os.path.join(dataset_dir, "metadata.csv")

# Download only if missing
if not os.path.exists(waveform_path):
    print("Downloading waveforms.hdf5...")
    gdown.download(waveform_url, waveform_path, quiet=False)
else:
    print("waveforms.hdf5 already exists in data/raw/, skipping download.")

if not os.path.exists(metadata_path):
    print("Downloading metadata.csv...")
    gdown.download(metadata_url, metadata_path, quiet=False)
else:
    print("metadata.csv already exists in data/raw/, skipping download.")

# Data Verification (A1 Requirements)
print("\n--- DATA VERIFICATION (A1) ---")
waveform_size_gb = os.path.getsize(waveform_path) / (1024 ** 3)
metadata_size_mb = os.path.getsize(metadata_path) / (1024 ** 2)
print(f"Waveform file size: {waveform_size_gb:.2f} GB")
print(f"Metadata file size: {metadata_size_mb:.2f} MB")

df_meta = pd.read_csv(metadata_path)
print(f"Total metadata records: {len(df_meta)}")
print("\nFirst record:")
print(df_meta.head(1))
print("\nLast record:")
print(df_meta.tail(1))

with h5py.File(waveform_path, "r") as hf:
    keys = list(hf.keys())
    print(f"\nTotal waveform HDF5 keys: {len(keys)}")
    if len(keys) > 0:
        print(f"First waveform key: {keys[0]}")
        print(f"Last waveform key: {keys[-1]}")
```
