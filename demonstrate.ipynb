```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# list containing plate boundary types
types = ['divergent', 'convergent', 'transform']

# Numpy array of distance in km
distances = np.array([0, 5, 10, 15, 20, 25, 30, 35, 40, 45, 50, 55, 60, 65, 70, 80, 85, 90, 95, 100])


dist_corr = np.array([1.4, 1.58, 1.72, 1.86, 1.98, 2.08, 2.18, 2.26, 2.34, 2.4, 2.47, 2.53, 2.6, 2.65, 2.7, 2.8, 2.86, 2.91, 2.96, 3.0])
```

```python
def km_to_m(x):
    return x * 1000

distances_m = km_to_m(distances)
```

```python
plt.plot(distances, dist_corr)
plt.xlabel('Distance (km)')
plt.ylabel('Distance Correction')
plt.title('Seismic Distance Correction vs Distances')
plt.show()
```
