```python
import os
import gdown

dataset_dir = "./pnwml/downloadedPNW"
os.makedirs(dataset_dir, exist_ok=True)

waveform_url = "https://drive.google.com/file/d/1SrbiQpBpU6mPq5Un_lJpPcfBekwczLzp/view?usp=share_link"
metadata_url = "https://drive.google.com/file/d/1HK2AuWPQj3dCdKShYcJ7a5E577XASrab/view?usp=share_link"

gdown.download(
    waveform_url,
    os.path.join(dataset_dir, "waveforms.hdf5"),
    quiet=False
)

gdown.download(
    metadata_url,
    os.path.join(dataset_dir, "metadata.csv"),
    quiet=False
)
```
